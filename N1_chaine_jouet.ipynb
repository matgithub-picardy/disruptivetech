<b style="color:#1F3A5F">Rappel du chapitre</b>

Une fonction de hachage transforme n'importe quel texte en une suite de 64 caractères, l'**empreinte**. Le même texte donne toujours la même empreinte. Changer un seul caractère change toute l'empreinte. On ne sait pas non plus retrouver le texte à partir de l'empreinte.

</div>



<div>

<span style="color:#4A7C59;font-weight:bold">&#9660; La cellule ci-dessous</span> crée une fonction <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">empreinte</code> qui calcule l'empreinte SHA-256 d'un texte, la fonction utilisée par Bitcoin. Le module <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">hashlib</code> de Python fait le calcul. Nous l'enveloppons simplement dans une fonction au nom plus parlant.

</div>

```python
import hashlib   # bibliothèque standard de Python qui contient SHA-256

def empreinte(texte):
    """Rend l'empreinte SHA-256 d'un texte, sous forme de 64 caractères hexadécimaux."""
    octets = texte.encode("utf-8")              # l'ordinateur hache des octets, pas des lettres
    return hashlib.sha256(octets).hexdigest()   # calcul de l'empreinte, écrite en hexadécimal (0-9, a-f)

print(empreinte("Alice paie 12 euros a Bob"))
print(empreinte("Alice paie 13 euros a Bob"))
```

<div style="background-color:#FFFFFF;color:#1E293B;font-family:'Helvetica Neue', Helvetica, Arial, sans-serif;line-height:1.55;padding:14px 26px 10px 26px">

Les deux phrases ne diffèrent que d'un chiffre et les empreintes n'ont rien en commun.

<span style="color:#4A7C59;font-weight:bold">&#9660; La cellule ci-dessous</span> mesure l'**effet avalanche** : combien de bits (les 0 et les 1 de l'empreinte) changent quand on modifie un caractère. Une empreinte compte 256 bits. Si le hachage est bon, environ la moitié, soit 128, doivent changer.

</div>

```python
def bits_differents(texte1, texte2):
    """Compte les bits qui diffèrent entre les empreintes de deux textes (sur 256)."""
    e1 = int(empreinte(texte1), 16)    # on convertit chaque empreinte en un grand nombre entier
    e2 = int(empreinte(texte2), 16)
    return bin(e1 ^ e2).count("1")     # ^ marque d'un 1 chaque bit différent, puis on compte les 1

print(bits_differents("Alice paie 12 euros a Bob", "Alice paie 13 euros a Bob"), "bits différents sur 256")
```

<h2 id="etape-4" style="color:#1F3A5F;font-family:'Helvetica Neue', Helvetica, Arial, sans-serif;border-bottom:2px solid #4A7C59;padding-bottom:4px;margin-top:4px">4. Chaîner des blocs et tenter une falsification</h2>
</div>

**Ce que vous allez faire.** Construire une petite chaîne de trois blocs, écrire un vérificateur puis jouer le rôle d'un fraudeur qui modifie un paiement passé.

<div style="border-left:4px solid #1F3A5F;background-color:#F1F3F6;color:#1E293B;padding:8px 16px 4px 16px;margin:16px 0 18px 0">

<b style="color:#1F3A5F">Rappel du chapitre</b>

Un bloc regroupe des transactions et contient l'empreinte du bloc précédent. Ce lien fait une **chaîne** : modifier un bloc ancien casse tous les liens qui suivent.

</div>

<span style="color:#4A7C59;font-weight:bold">&#9660; La cellule ci-dessous</span> définit deux fonctions.
- <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">empreinte_bloc(bloc)</code> calcule l'empreinte d'un bloc à partir de tout son contenu, sauf la rubrique <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">"empreinte"</code> elle-même.
- <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">creer_bloc(...)</code> fabrique un bloc, c'est-à-dire un dictionnaire avec un numéro, l'empreinte du bloc précédent, une liste de transactions et un <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">nonce</code> (un nombre qui servira au minage à l'étape 5). Elle y range ensuite sa propre empreinte.

</div>

```python
import json   # pour transformer un dictionnaire en texte, toujours écrit de la même façon

def empreinte_bloc(bloc):
    """Empreinte de tout le contenu du bloc, sauf la rubrique 'empreinte'."""
    contenu = {cle: valeur for cle, valeur in bloc.items() if cle != "empreinte"}
    texte = json.dumps(contenu, sort_keys=True, ensure_ascii=False)   # même contenu => même texte
    return empreinte(texte)

def creer_bloc(numero, empreinte_precedente, transactions, nonce=0):
    """Fabrique un bloc et y range sa propre empreinte."""
    bloc = {
        "numero": numero,                       # position du bloc dans la chaîne
        "precedent": empreinte_precedente,      # le lien avec le bloc d'avant
        "transactions": transactions,           # ce que le bloc enregistre
        "nonce": nonce,                         # nombre libre, utilisé pour miner (étape 5)
    }
    bloc["empreinte"] = empreinte_bloc(bloc)
    return bloc
```

<span style="color:#4A7C59;font-weight:bold">&#9660; La cellule ci-dessous</span> construit une chaîne de trois blocs. Le premier bloc, dit bloc de genèse, n'a pas de prédécesseur : on met 64 zéros à la place. Chaque bloc suivant reçoit l'empreinte de celui d'avant.

</div>

```python
bloc0 = creer_bloc(0, "0" * 64, ["Création de 100 pièces pour Alice"])
bloc1 = creer_bloc(1, bloc0["empreinte"], ["Alice paie 12 à Bob"])
bloc2 = creer_bloc(2, bloc1["empreinte"], ["Bob paie 5 à Chloé"])
chaine = [bloc0, bloc1, bloc2]

for bloc in chaine:
    print(f"Bloc {bloc['numero']} | précédent : {bloc['precedent'][:12]}... | empreinte : {bloc['empreinte'][:12]}...")
```

<div style="background-color:#FFFFFF;color:#1E293B;font-family:'Helvetica Neue', Helvetica, Arial, sans-serif;line-height:1.55;padding:14px 26px 10px 26px">

Vérifiez à l'écran que la rubrique *précédent* de chaque bloc reprend l'empreinte du bloc du dessus. On n'affiche que les 12 premiers caractères pour la lisibilité.

N'importe qui peut vérifier une chaîne en refaisant deux contrôles sur chaque bloc, sans faire confiance à personne.

<span style="color:#4A7C59;font-weight:bold">&#9660; La cellule ci-dessous</span> crée la fonction <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">verifier_chaine</code>. Elle fait ces deux contrôles :
1. l'empreinte rangée dans le bloc correspond à son contenu actuel,
2. la rubrique *précédent* reprend bien l'empreinte du bloc d'avant.

</div>

```python
def verifier_chaine(chaine):
    """Rend (True, message) si la chaîne est intacte, (False, message) sinon."""
    for i, bloc in enumerate(chaine):
        # Contrôle 1 : le contenu du bloc a changé depuis le calcul de son empreinte ?
        if bloc["empreinte"] != empreinte_bloc(bloc):
            return False, f"le bloc {i} a été modifié (son empreinte ne correspond plus)"
        # Contrôle 2 : le lien avec le bloc précédent est-il intact ?
        if i > 0 and bloc["precedent"] != chaine[i - 1]["empreinte"]:
            return False, f"le lien entre le bloc {i-1} et le bloc {i} est cassé"
    return True, "chaîne valide"

print(verifier_chaine(chaine))
```

</div>

**Ce que vous allez faire.** Programmer le minage : chercher par essais successifs un nonce qui donne une empreinte commençant par des zéros, puis mesurer comment le nombre d'essais augmente avec la difficulté.

<div style="border-left:4px solid #1F3A5F;background-color:#F1F3F6;color:#1E293B;padding:8px 16px 4px 16px;margin:16px 0 18px 0">

<b style="color:#1F3A5F">Rappel du chapitre</b>

Pour qu'un bloc soit accepté, son empreinte doit commencer par un certain nombre de zéros, la **difficulté**. Le seul moyen d'y arriver est d'essayer des valeurs du <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">nonce</code> une par une, jusqu'à tomber sur une empreinte qui convient. C'est long à trouver et immédiat à vérifier. C'est ce que fait le bouton *Mine* du site de l'étape 2.

</div>

<span style="color:#4A7C59;font-weight:bold">&#9660; La cellule ci-dessous</span> crée la fonction <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">miner</code>. Elle part de <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">nonce = 0</code>, calcule l'empreinte et augmente le nonce de 1 tant que l'empreinte ne commence pas par assez de zéros. Elle rend le bloc trouvé et le nombre d'essais.

</div>

```python
def miner(numero, empreinte_precedente, transactions, difficulte):
    """Cherche un nonce qui donne une empreinte commençant par `difficulte` zéros."""
    cible = "0" * difficulte          # par exemple "000" pour une difficulté de 3
    nonce = 0
    while True:                       # on essaie tant qu'on n'a pas trouvé
        bloc = creer_bloc(numero, empreinte_precedente, transactions, nonce)
        if bloc["empreinte"].startswith(cible):
            return bloc, nonce + 1    # nombre d'essais = nonce + 1 (on a commencé à 0)
        nonce += 1                    # raté : on passe au nonce suivant

for d in range(1, 5):
    bloc, essais = miner(1, bloc0["empreinte"], ["Alice paie 12 à Bob"], d)
    print(f"difficulté {d} : {essais:>6} essais, empreinte {bloc['empreinte'][:16]}...")
```

Chaque caractère hexadécimal peut prendre 16 valeurs. Exiger un zéro de plus divise donc par 16 la chance de réussir à chaque essai. En moyenne, il faut 16 essais pour un zéro, 256 pour deux, 4 096 pour trois, 65 536 pour quatre. Un seul tirage peut s'écarter beaucoup de cette moyenne : c'est une loterie.

<div style="border:1px solid #B45309;background-color:#FFFFFF;color:#1E293B;padding:0 16px 4px 16px;margin:18px 0 18px 0">

Comparez le nombre d'essais à la moyenne de 16⁵ = 1 048 576. Que se passerait-il pour un fraudeur qui devrait refaire ce calcul pour plusieurs blocs ?

</div>

</div>

```python
import time
debut = time.time()
# À COMPLÉTER : remplacez ... par la difficulté 5
bloc, essais = miner(1, bloc0["empreinte"], ["Alice paie 12 à Bob"], ...)
print(f"{essais} essais en {time.time() - debut:.1f} secondes")
```

**Ce que vous allez faire.** Reproduire en Python une signature simplifiée et un registre de pièces, puis voir ce qui se passe quand Alice signe deux paiements avec la même pièce.

<div style="border-left:4px solid #1F3A5F;background-color:#F1F3F6;color:#1E293B;padding:8px 16px 4px 16px;margin:16px 0 18px 0">

<b style="color:#1F3A5F">Rappel du chapitre</b>

Pour dépenser une pièce, il faut la **signer** avec sa clé privée. N'importe qui peut vérifier la signature avec la clé publique. Dans Bitcoin, on ne tient pas de soldes. On possède des **pièces** (les UTXO, sorties de transaction non dépensées) qui se dépensent en entier et une seule fois, à la manière de billets.

</div>

<span style="color:#4A7C59;font-weight:bold">&#9660; La cellule ci-dessous</span> fabrique une signature simplifiée sur le principe de RSA. La clé privée est un nombre secret <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">d</code> et la clé publique est le couple <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">(e, n)</code>. Signer revient à transformer l'empreinte du message avec la clé privée. Vérifier revient à refaire la transformation inverse avec la clé publique. Les nombres sont trop petits pour être sûrs : la signature sert à comprendre le principe et ne protège rien.

</div>

```python
def creer_cles(p, q):
    """Fabrique une paire de clés simplifiée à partir de deux nombres premiers p et q."""
    n = p * q
    e = 65537                             # partie publique, la même pour tout le monde
    d = pow(e, -1, (p - 1) * (q - 1))     # partie privée, calculable seulement si on connaît p et q
    return {"privee": d, "publique": (e, n)}

def signer(message, cle_privee, cle_publique):
    h = int(empreinte(message), 16) % cle_publique[1]   # l'empreinte du message, ramenée sous n
    return pow(h, cle_privee, cle_publique[1])          # transformation avec la clé privée

def verifier_signature(message, signature, cle_publique):
    e, n = cle_publique
    h = int(empreinte(message), 16) % n
    return pow(signature, e, n) == h                    # la transformation inverse doit redonner l'empreinte

cles_alice = creer_cles(1000003, 1000033)
s = signer("Alice paie 7 à Bob", cles_alice["privee"], cles_alice["publique"])
print("signature valide ?", verifier_signature("Alice paie 7 à Bob", s, cles_alice["publique"]))
print("message modifié, signature valide ?", verifier_signature("Alice paie 70 à Bob", s, cles_alice["publique"]))
```

<div style="background-color:#FFFFFF;color:#1E293B;font-family:'Helvetica Neue', Helvetica, Arial, sans-serif;line-height:1.55;padding:14px 26px 10px 26px">

La signature ne vaut que pour le message signé. Changer le montant la rend invalide, exactement comme sur le site à l'étape 6.

<span style="color:#4A7C59;font-weight:bold">&#9660; La cellule ci-dessous</span> met en place le registre des pièces. Le dictionnaire <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">utxo</code> contient les pièces non dépensées : un identifiant, un.e propriétaire et un montant. La fonction <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">appliquer</code> vérifie une transaction puis met le registre à jour. Elle fait trois contrôles, dans cet ordre :
1. chaque pièce dépensée existe encore dans le registre (sinon, elle a déjà été dépensée),
2. la signature est celle du ou de la propriétaire des pièces,
3. on ne crée pas plus que ce qu'on dépense.

</div>

```python
cles = {"Alice": cles_alice, "Bob": creer_cles(1000037, 1000039), "Charlie": creer_cles(1000081, 1000099)}
utxo = {"piece1": {"proprietaire": "Alice", "montant": 10}}   # au départ, Alice possède une pièce de 10

def message_de(transaction):
    """Le texte qui est signé : les pièces dépensées et les bénéficiaires."""
    return json.dumps({"entrees": transaction["entrees"], "sorties": transaction["sorties"]}, sort_keys=True)

def appliquer(transaction, utxo):
    for piece in transaction["entrees"]:
        if piece not in utxo:                                             # contrôle 1
            return f"REFUSÉE : la pièce {piece} n'existe pas ou a déjà été dépensée"
        proprio = utxo[piece]["proprietaire"]
        if not verifier_signature(message_de(transaction), transaction["signature"], cles[proprio]["publique"]):
            return "REFUSÉE : signature invalide"                         # contrôle 2
    total_entrees = sum(utxo[p]["montant"] for p in transaction["entrees"])
    total_sorties = sum(montant for _, montant in transaction["sorties"])
    if total_sorties > total_entrees:                                     # contrôle 3
        return "REFUSÉE : on ne peut pas créer de la monnaie"
    for piece in transaction["entrees"]:                                  # les pièces dépensées disparaissent
        del utxo[piece]
    for k, (beneficiaire, montant) in enumerate(transaction["sorties"]):  # de nouvelles pièces sont créées
        utxo[f"{transaction['id']}:{k}"] = {"proprietaire": beneficiaire, "montant": montant}
    frais = round(total_entrees - total_sorties, 2)   # arrondi pour l'affichage
    return f"ACCEPTÉE (frais pour le mineur : {frais})"

def creer_transaction(id_tx, signataire, entrees, sorties):
    tx = {"id": id_tx, "entrees": entrees, "sorties": sorties}
    tx["signature"] = signer(message_de(tx), cles[signataire]["privee"], cles[signataire]["publique"])
    return tx
```

<div style="background-color:#FFFFFF;color:#1E293B;font-family:'Helvetica Neue', Helvetica, Arial, sans-serif;line-height:1.55;padding:14px 26px 10px 26px">

**La double dépense.** Alice signe deux transactions qui dépensent la même pièce, l'une vers Bob et l'autre vers Charlie. Les deux signatures sont valides. Le registre accepte la première arrivée et refuse la seconde parce que la pièce n'existe plus.

</div>

```python
tx_bob = creer_transaction("tx1", "Alice", ["piece1"], [("Bob", 7), ("Alice", 2.9)])        # 0,1 de frais
tx_charlie = creer_transaction("tx2", "Alice", ["piece1"], [("Charlie", 9.9)])

print("vers Bob     :", appliquer(tx_bob, utxo))
print("vers Charlie :", appliquer(tx_charlie, utxo))
print("pièces non dépensées :", utxo)
```

**Ce que vous allez faire.** Jouer au minage avec des dés virtuels. Cinq équipes de mineurs, de A à E, lancent des paires de dés. Une paire de dés lancée représente un essai de hachage. Une paire qui fait 11 ou 12 trouve un bloc. Les équipes n'ont pas le même nombre de paires : c'est leur puissance de calcul. A, B et C ont une paire, D en a deux et E en a trois.

<div style="border-left:4px solid #1F3A5F;background-color:#F1F3F6;color:#1E293B;padding:8px 16px 4px 16px;margin:16px 0 18px 0">

<b style="color:#1F3A5F">Rappel du chapitre</b>

Dans le consensus de Nakamoto, chaque bloc est attribué par une loterie dont les chances sont proportionnelles à la puissance de calcul. Les autres mineurs acceptent le bloc en construisant le suivant dessus. Quand deux blocs sont trouvés en même temps, la chaîne se divise : c'est une **fourche**. Le réseau finit par ne garder qu'une branche.

```python
# ---- Préparation du jeu (à exécuter une fois, inutile de lire ce code) ----
import random, time

PUISSANCE = {"A": 1, "B": 1, "C": 1, "D": 2, "E": 3}   # nombre de paires de dés de chaque équipe

def lancer_paires(n):
    """Lance n paires de dés."""
    return [(random.randint(1, 6), random.randint(1, 6)) for _ in range(n)]

def montrer(equipe, paires, cible):
    """Affiche les dés d'une équipe. Rend True si une paire atteint la cible (bloc trouvé)."""
    textes, trouve = [], False
    for a, b in paires:
        ok = a + b >= cible
        trouve = trouve or ok
        textes.append(f"{a}+{b}={a+b:<2}{' ✔' if ok else '  '}")
    print(f"   Équipe {equipe} : " + "   ".join(textes))
    return trouve

def nouvelle_partie():
    global m1, m2
    m1 = {"tour": 0, "chaine": []}
    m2 = {"tour": 0, "publique": 1, "secrete": 0, "livre": False, "fini": False}

def tour_manche_1(nb_tours=15):
    if m1["tour"] >= nb_tours:
        print("La manche 1 est finie. Pour rejouer, exécutez nouvelle_partie() puis relancez cette cellule.")
        return
    m1["tour"] += 1
    print(f"Tour {m1['tour']} sur {nb_tours}. Un bloc est trouvé avec une paire de 11 ou 12.\n")
    gagnants = [e for e, n in PUISSANCE.items() if montrer(e, lancer_paires(n), 11)]
    print()
    if not gagnants:
        print("   Aucun bloc à ce tour.")
    else:
        retenu = random.choice(gagnants)
        m1["chaine"].append(retenu)
        if len(gagnants) > 1:
            print(f"   Fourche : {' et '.join(gagnants)} trouvent un bloc en même temps. Le réseau garde celui de {retenu}.")
        else:
            print(f"   L'équipe {retenu} trouve le bloc.")
    print("   Chaîne : " + (" - ".join(m1["chaine"]) or "vide"))
    if m1["tour"] == nb_tours:
        total_des, total_blocs = sum(PUISSANCE.values()), max(len(m1["chaine"]), 1)
        print(f"\nFin de la manche 1 : {len(m1['chaine'])} blocs en {nb_tours} tours.\n")
        print("   Équipe   part des dés   part des blocs")
        for e, n in PUISSANCE.items():
            print(f"     {e}        {100*n/total_des:5.0f} %        {100*m1['chaine'].count(e)/total_blocs:5.0f} %")

def tour_manche_2(confirmations=2, max_tours=25):
    if m2["fini"]:
        print("La manche 2 est finie. Pour rejouer, exécutez nouvelle_partie() puis relancez cette cellule.")
        return
    m2["tour"] += 1
    print(f"Tour {m2['tour']}. A, B, C et D minent la chaîne publique. E mine en secret.\n")
    honnetes = [e for e in "ABCD" if montrer(e, lancer_paires(PUISSANCE[e]), 11)]
    attaque = montrer("E", lancer_paires(PUISSANCE["E"]), 11)
    m2["publique"] += 1 if honnetes else 0
    m2["secrete"] += 1 if attaque else 0
    print("\n   Chaîne publique   : P" + " - H" * (m2["publique"] - 1) + "    (P contient le paiement de E à A)")
    print("   Branche secrète E : " + (" - ".join(["E"] * m2["secrete"]) or "vide"))
    if not m2["livre"] and m2["publique"] >= 1 + confirmations:
        m2["livre"] = True
        print(f"\n   A voit {confirmations} confirmations après le bloc P et livre le bien.")
    if m2["livre"] and m2["secrete"] > m2["publique"]:
        m2["fini"] = True
        print("\n   L'ATTAQUE RÉUSSIT. La branche secrète est plus longue. E la publie, le réseau la suit")
        print("   et le paiement à A disparaît. A a livré le bien sans être payée.")
    elif m2["tour"] >= max_tours:
        m2["fini"] = True
        print(f"\n   E abandonne après {max_tours} tours : l'attaque échoue. Le minage secret a coûté pour rien.")

def manche_3():
    double = {e: 2 * n for e, n in PUISSANCE.items()}
    for phase, cible, nb in (("avant l'ajustement, bloc avec 11 ou 12", 11, 5), ("après l'ajustement, bloc avec un double six", 12, 10)):
        print(f"\nChaque équipe a doublé ses dés, {phase}.")
        blocs = fourches = 0
        for t in range(1, nb + 1):
            gagnants = [e for e, n in double.items() if any(a + b >= cible for a, b in lancer_paires(n))]
            blocs += 1 if gagnants else 0
            fourches += 1 if len(gagnants) > 1 else 0
            etat = "aucun bloc" if not gagnants else ("FOURCHE entre " + " et ".join(gagnants) if len(gagnants) > 1 else "bloc de " + gagnants[0])
            print(f"   tour {t:>2} : {etat}")
            time.sleep(0.4)
        print(f"   Bilan : {blocs} blocs en {nb} tours, dont {fourches} tours avec fourche.")

nouvelle_partie()
print("Le jeu est prêt.")
```

<h2 id="etape-9" style="color:#1F3A5F;font-family:'Helvetica Neue', Helvetica, Arial, sans-serif;border-bottom:2px solid #4A7C59;padding-bottom:4px;margin-top:4px">9. Simuler l'attaque de double dépense</h2>
</div>

**Ce que vous allez faire.** Calculer la probabilité de réussite de l'attaque de la manche 2 selon la part de calcul de l'attaquant.e et le nombre de confirmations, d'abord par simulation, puis avec la formule de Nakamoto.

<div style="border-left:4px solid #1F3A5F;background-color:#F1F3F6;color:#1E293B;padding:8px 16px 4px 16px;margin:16px 0 18px 0">

<b style="color:#1F3A5F">Rappel du chapitre</b>

Un.e attaquant.e qui détient une part <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">q</code> du calcul paie Bob, attend que Bob voie <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">z</code> confirmations et livre le bien, puis mine en secret une branche concurrente. L'attaque réussit si la branche secrète rattrape la chaîne publique.

</div>

<span style="color:#4A7C59;font-weight:bold">&#9660; La cellule ci-dessous</span> joue un grand nombre de fois, avec le hasard, la course entre l'attaquant.e et les mineurs honnêtes. C'est une simulation de **Monte-Carlo**. Chaque nouveau bloc revient à l'attaquant.e avec la probabilité <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">q</code> et aux mineurs honnêtes avec la probabilité <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">1 - q</code>. La fréquence des succès estime la probabilité de réussite.

</div>

```python
import random, math

def une_attaque(q, z):
    """Joue une course. Rend True si l'attaquant.e rattrape la chaîne publique."""
    honnetes, attaquant = 0, 0
    while honnetes < z:                 # 1re phase : Bob attend z blocs honnêtes
        if random.random() < q:
            attaquant += 1              # bloc trouvé en secret par l'attaquant.e
        else:
            honnetes += 1
    retard = z - attaquant              # 2e phase : course à partir de ce retard
    while 0 < retard < 60:              # au-delà de 60 blocs de retard, on considère l'attaque perdue
        retard += -1 if random.random() < q else 1
    return retard <= 0

def proba_simulee(q, z, essais=20000):
    return sum(une_attaque(q, z) for _ in range(essais)) / essais

def proba_nakamoto(q, z):
    """Formule de Nakamoto (2008, section 11)."""
    p = 1 - q
    lam = z * q / p
    echec = sum(math.exp(-lam) * lam**k / math.factorial(k) * (1 - (q / p)**(z - k)) for k in range(z + 1))
    return 1 - echec

for z in (1, 2, 4, 6):
    print(f"q = 0,30 et z = {z} : simulation {proba_simulee(0.3, z):.3f}, formule {proba_nakamoto(0.3, z):.3f}")
```

<div style="background-color:#FFFFFF;color:#1E293B;font-family:'Helvetica Neue', Helvetica, Arial, sans-serif;line-height:1.55;padding:14px 26px 10px 26px">

Les deux colonnes sont proches sans être identiques. La formule de Nakamoto approche le nombre de blocs secrets par une loi de Poisson alors que la simulation joue la course exactement. Les deux probabilités décroissent rapidement quand <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">z</code> augmente.

<span style="color:#4A7C59;font-weight:bold">&#9660; La cellule ci-dessous</span> trace la probabilité de réussite selon le nombre de confirmations, pour plusieurs parts de calcul. L'échelle verticale est logarithmique : une droite signifie une décroissance exponentielle.

</div>

```python
import matplotlib.pyplot as plt

COULEURS = ["#1F3A5F", "#4A7C59", "#B45309", "#9F1239"]   # couleurs du cours

fig, ax = plt.subplots(figsize=(7, 4))
zs = list(range(0, 11))
for q, couleur in zip((0.1, 0.2, 0.3, 0.4), COULEURS):
    ax.semilogy(zs, [proba_nakamoto(q, z) for z in zs], marker="o", color=couleur, label=f"q = {q}")
ax.set_xlabel("confirmations attendues par Bob (z)")
ax.set_ylabel("probabilité de réussite de l'attaque")
ax.set_title("Attaque de double dépense : formule de Nakamoto", color="#1F3A5F", loc="left")
ax.spines[["top", "right"]].set_visible(False)
ax.grid(True, which="both", alpha=0.25)
ax.legend(frameon=False)
plt.show()
```
