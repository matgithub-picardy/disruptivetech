**Ce que vous allez faire.** Calculer la recette d'un bloc de Bitcoin puis la valeur qu'elle protège contre une attaque majoritaire selon la condition de Budish (2025). Vous verrez comment cette valeur dépend du prix du bitcoin, de la durée de l'attaque et de la subvention, divisée par deux tous les quatre ans.

<div style="border-left:4px solid #1F3A5F;background-color:#F1F3F6;color:#1E293B;padding:8px 16px 4px 16px;margin:16px 0 18px 0">

<b style="color:#1F3A5F">Rappel du chapitre</b>

La **recette d'un bloc** vaut <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">R_bloc = (subvention + frais) × prix du bitcoin</code>. Une attaque majoritaire de <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">t</code> blocs coûte environ <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">α × t × R_bloc</code> si l'attaquant.e loue autant de calcul que le reste du réseau (<code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">α = 1</code>). L'attaque n'est pas rentable si ce qu'elle rapporte, <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">V_attaque</code>, reste inférieur à ce coût. La valeur protégée vaut donc au plus <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">V_max = α × t × R_bloc</code>. 

Le paramètre <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">t</code> est la durée de l'attaque, comptée en blocs. Pour annuler un paiement déjà reçu, l'attaquant.e doit produire en secret une branche plus longue que la chaîne publique. Il.elle doit garder la majorité du calcul pendant tout le temps où la personne payée attend des confirmations avant de livrer le bien. Plus cette attente est longue, plus l'attaque dure et plus elle coûte : à chaque bloc, l'attaquant.e paie pour louer un calcul dont le prix est de l'ordre de la recette d'un bloc. Le protocole produit un bloc toutes les 10 minutes en moyenne. Une heure correspond ainsi à 60 / 10 = 6 blocs, soit <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">t = 6</code>. Ces six blocs correspondent aux six confirmations souvent recommandées avant de considérer un paiement important comme définitif. Un jour correspond à 24 × 6 = 144 blocs, soit <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">t = 144</code>. Cette durée est celle d'une attaque contre une personne qui attendrait une journée avant de livrer.

</div>

<span style="color:#4A7C59;font-weight:bold">&#9660; La cellule ci-dessous</span> crée deux fonctions, l'une pour la recette d'un bloc et l'autre pour la valeur protégée, puis reproduit le tableau de la Section 1.3 du chapitre avec les chiffres de novembre 2026.

</div>

```python
def recette_bloc(subvention, frais, prix):
    """Recette d'un bloc en dollars : (bitcoins neufs + frais en bitcoins) × prix du bitcoin."""
    return (subvention + frais) * prix

def valeur_protegee(p_bloc, t, alpha=1):
    """Valeur maximale qu'une attaque de t blocs ne rend pas rentable (condition de Budish)."""
    return alpha * t * p_bloc

P = recette_bloc(subvention=3.125, frais=0.02, prix=74_000)
print(f"Recette d'un bloc : {P:,.0f} dollars".replace(",", " "))
print()
for nom, t in (("1 heure", 6), ("1 jour", 144), ("1 semaine", 1008)):
    print(f"Attaque de {nom:<9} : valeur protégée {valeur_protegee(P, t) / 1e6:7.1f} M$,"
          f"  recette par bloc nécessaire pour protéger 1 Md$ : {1e9 / t / 1e6:6.1f} M$")
```

<h2 id="etape-2" style="color:#1F3A5F;font-family:'Helvetica Neue', Helvetica, Arial, sans-serif;border-bottom:2px solid #4A7C59;padding-bottom:4px;margin-top:4px">2. Simuler la libre entrée du minage</h2>
</div>

**Ce que vous allez faire.** Simuler la réponse de la puissance de calcul à une hausse puis à une baisse du prix du bitcoin. Le modèle est une simplification construite pour l'exercice : ses paramètres ne sont pas estimés sur des données.

<div style="border-left:4px solid #1F3A5F;background-color:#F1F3F6;color:#1E293B;padding:8px 16px 4px 16px;margin:16px 0 18px 0">

<b style="color:#1F3A5F">Rappel du chapitre</b>

Avec la **libre entrée**, les mineurs entrent tant que la recette d'un bloc dépasse le coût du calcul. L'entrée s'arrête quand <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">N × c = R_bloc</code>, où <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">N</code> est le nombre d'unités de calcul et <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">c</code> le coût d'une unité pendant la durée d'un bloc. Dans le modèle de Prat et Walter (2021), les machines (ASIC) ne servent qu'au minage et leur achat est irréversible. La puissance de calcul réagit au prix avec retard et de façon amortie.

</div>

<div style="border-left:4px solid #475569;background-color:#F3F4F6;color:#1E293B;padding:8px 16px 4px 16px;margin:16px 0 18px 0">

<b style="color:#475569">Les règles du modèle simplifié</b>

On suit la puissance de calcul mois par mois, en indice (100 au départ). Le coût <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">c</code> d'une unité comprend l'électricité (60 %) et l'usure des machines et le capital immobilisé (40 %).

1. **Entrée.** Si la recette dépasse le coût total du calcul, des machines sont commandées. Leur fabrication et leur livraison prennent du temps : chaque mois, la puissance comble 10 % de l'écart avec le niveau de libre entrée <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">R_bloc / c</code>.
2. **Sortie.** Une machine déjà achetée continue de tourner tant que la recette couvre l'électricité, même si elle ne couvre plus l'usure et le capital. Les machines s'arrêtent seulement si la recette par unité tombe sous le coût de l'électricité.
3. **Prix.** Le prix du bitcoin augmente de 50 % au mois 6 puis tombe à 80 % de son niveau initial au mois 30.

</div>

<span style="color:#4A7C59;font-weight:bold">&#9660; La cellule ci-dessous</span> simule les 48 mois et trace la puissance de calcul, son niveau de libre entrée et la recette par unité de calcul rapportée au coût <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">c</code>.

</div>

```python
import matplotlib.pyplot as plt
COULEURS = ["#1F3A5F", "#4A7C59", "#B45309", "#9F1239"]   # couleurs du cours

def simuler_minage(mois=48, choc=1.5, vitesse=0.10, part_electricite=0.6):
    """Puissance de calcul (indice 100 au départ) mois par mois."""
    c = 1.0                    # coût d'une unité de calcul (unités choisies pour que N = 100 au départ)
    N = 100.0
    puissance, libre_entree, recette_unite = [], [], []
    for m in range(mois):
        prix = 1.0 if m < 6 else (choc if m < 30 else 0.8)   # prix du bitcoin, en indice
        recette = 100 * prix                          # recette d'un bloc, proportionnelle au prix
        N_cible = recette / c                         # niveau de libre entrée : N × c = R_bloc
        if N < N_cible:                               # entrée : des machines arrivent avec retard
            N += vitesse * (N_cible - N)
        elif recette / N < part_electricite * c:      # sortie : seulement si l'électricité n'est plus couverte
            N += vitesse * (recette / (part_electricite * c) - N)
        puissance.append(N)
        libre_entree.append(N_cible)
        recette_unite.append(recette / N / c)
    return puissance, libre_entree, recette_unite

puissance, libre_entree, recette_unite = simuler_minage()
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 3.8))
ax1.plot(puissance, color=COULEURS[0], label="puissance de calcul N")
ax1.plot(libre_entree, color=COULEURS[2], linestyle="--", label="niveau de libre entrée R_bloc / c")
ax1.set_xlabel("mois"); ax1.set_ylabel("indice (100 au départ)")
ax1.set_title("Puissance de calcul", color="#1F3A5F", loc="left")
ax2.plot(recette_unite, color=COULEURS[1])
ax2.axhline(1, color="grey", linewidth=0.8)
ax2.set_xlabel("mois"); ax2.set_ylabel("recette par unité / coût c")
ax2.set_title("Rentabilité du calcul", color="#1F3A5F", loc="left")
for ax in (ax1, ax2):
    ax.spines[["top", "right"]].set_visible(False)
    ax.grid(True, alpha=0.25)
ax1.legend(frameon=False, fontsize=8)
plt.show()
```

<div style="background-color:#FFFFFF;color:#1E293B;font-family:'Helvetica Neue', Helvetica, Arial, sans-serif;line-height:1.55;padding:14px 26px 10px 26px">

Lisez les deux graphiques ensemble. Après la hausse du prix, la recette par unité dépasse le coût : les mineurs déjà installés touchent une rente temporaire. La rente disparaît à mesure que de nouvelles machines arrivent. Après la baisse du prix, la puissance diminue un peu mais reste bien au-dessus de son nouveau niveau de libre entrée. Les machines déjà payées continuent de tourner tant qu'elles couvrent l'électricité, avec une recette par unité inférieure au coût total.

<div style="border-left:4px solid #1F3A5F;background-color:#F1F3F6;color:#1E293B;padding:8px 16px 4px 16px;margin:16px 0 18px 0">

<b style="color:#1F3A5F">Rappel du chapitre</b>

Un mineur qui détient une part <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">θ</code> du calcul trouve en moyenne <code style="color:#9F1239;background-color:#F1F5F9;padding:1px 5px;border-radius:3px;font-size:0.92em">θ × 52 560</code> blocs par an. Le nombre de blocs qu'il trouve suit une **loi de Poisson**, la loi du nombre d'événements rares et indépendants pendant une durée donnée. Dans un **pool**, les mineurs mettent leur calcul en commun et se partagent les récompenses en proportion du calcul apporté, moyennant une commission versée au gestionnaire du pool.

</div>

<span style="color:#4A7C59;font-weight:bold">&#9660; La cellule ci-dessous</span> simule 10 000 années possibles pour le même mineur, seul puis dans un pool qui prélève 2 % de commission. Une récompense vaut la recette d'un bloc calculée à l'étape 1. Le coût annuel du mineur (électricité et machines) est fixé, pour l'exemple, à 80 % de son revenu moyen.

</div>

```python
import math, random
import matplotlib.pyplot as plt
COULEURS = ["#1F3A5F", "#4A7C59", "#B45309", "#9F1239"]   # couleurs du cours

def tirage_poisson(moyenne):
    """Nombre aléatoire d'événements selon une loi de Poisson (méthode de Knuth)."""
    seuil, k, p = math.exp(-moyenne), 0, 1.0
    while True:
        p *= random.random()
        if p < seuil:
            return k
        k += 1

theta = 0.00001                       # 0,001 % du calcul
blocs_attendus = theta * 52_560       # blocs trouvés en moyenne en un an
revenu_moyen = blocs_attendus * P     # P : recette d'un bloc (étape 1)
cout_annuel = 0.8 * revenu_moyen

seul = [tirage_poisson(blocs_attendus) * P for _ in range(10_000)]
pool = [0.98 * revenu_moyen] * 10_000   # le pool lisse le revenu, moins 2 % de commission

print(f"Blocs attendus en un an : {blocs_attendus:.2f}")
print(f"Revenu moyen seul : {sum(seul) / len(seul):,.0f} $".replace(",", " "))
print(f"Revenu moyen en pool : {pool[0]:,.0f} $".replace(",", " "))
print(f"Années en perte : seul {100 * sum(r < cout_annuel for r in seul) / len(seul):.0f} %, en pool 0 %")

fig, ax = plt.subplots(figsize=(7, 3.8))
ax.hist([r / 1000 for r in seul], bins=30, color=COULEURS[0], alpha=0.8, label="seul")
ax.axvline(pool[0] / 1000, color=COULEURS[2], linewidth=2.5, label="dans un pool")
ax.axvline(cout_annuel / 1000, color=COULEURS[3], linestyle="--", label="coût annuel")
ax.set_xlabel("revenu de l'année (milliers de dollars)")
ax.set_ylabel("nombre d'années simulées")
ax.set_title("Revenu annuel d'un mineur qui détient 0,001 % du calcul", color="#1F3A5F", loc="left")
ax.spines[["top", "right"]].set_visible(False)
ax.legend(frameon=False)
plt.show()
```

**Manche 1, enchère au premier prix.** Chacun.e fait une offre secrète. Les 4 offres les plus élevées passent dans le bloc 1, les 4 suivantes dans le bloc 2 et ainsi de suite. Chacun.e paie son offre.

**Manche 2, règle EIP-1559 simplifiée.** Avant chaque bloc, un frais de base est affiché. Il commence à 2 points. Vous décidez si vous l'acceptez. S'il y a plus de candidat.es que de places, un pourboire départage les candidat.es : il monte d'un point à la fois jusqu'à ce qu'il ne reste que 4 candidat.es. Le frais de base est détruit et le pourboire revient au producteur du bloc. Après un bloc plein, le frais de base est multiplié par 1,25. Après un bloc rempli à moitié ou moins, il est multiplié par 0,8. Le vrai protocole ajuste de 12,5 % au plus par bloc : le jeu accélère l'ajustement pour que l'effet soit visible en 6 blocs.

**Manche 3, choc de demande.** On rejoue la manche 2 avec 10 cartes NFT supplémentaires.

En manches 2 et 3, ils et elles acceptent le frais de base s'il leur laisse un gain et montent le pourboire jusqu'à une partie de ce gain.

```python
# ---- Préparation du jeu (à exécuter une fois, inutile de lire ce code) ----
import random

TYPES = {"A": (40, 8, 3), "B": (20, 4, 6), "C": (12, 2, 9), "D": (8, 1, 6), "E": (3, 0.5, 6), "F": (30, 6, 10)}
PLACES, BLOCS = 4, 6
RESULTATS = {}

paquet = [t for t, (v, w, n) in TYPES.items() if t != "F" for _ in range(n)]
ma_carte = random.choice(paquet)
v_moi, w_moi = TYPES[ma_carte][0], TYPES[ma_carte][1]

def distribuer(choc=False):
    """Les cartes des autres utilisateur.rices, plus la vôtre (nom 'vous')."""
    joueurs, deja_pris = [], False
    for t, (v, w, n) in TYPES.items():
        if t == "F" and not choc:
            continue
        for i in range(n):
            if t == ma_carte and not deja_pris:
                joueurs.append({"nom": "vous", "type": t, "v": v, "w": w}); deja_pris = True
            else:
                joueurs.append({"nom": f"{t}{i + 1}", "type": t, "v": v, "w": w})
    for j in joueurs:
        j.update(bloc=None, base=0, pourboire=0)
    return joueurs

def gain(j):
    return j["v"] - j["w"] * (j["bloc"] - 1) - j["base"] - j["pourboire"] if j["bloc"] else 0

def bilan(joueurs, nom):
    inclus = [j for j in joueurs if j["bloc"]]
    petits = [j for j in joueurs if j["type"] in "DE"]
    r = {"recette du producteur": sum(j["pourboire"] for j in inclus),
         "frais détruits": sum(j["base"] for j in inclus),
         "délai moyen (blocs)": sum(j["bloc"] - 1 for j in inclus) / max(len(inclus), 1),
         "petits paiements servis (%)": 100 * sum(1 for j in petits if j["bloc"]) / len(petits),
         "surplus total": sum(j["v"] - j["w"] * (j["bloc"] - 1) for j in inclus),
         "votre gain": gain(next(j for j in joueurs if j["nom"] == "vous"))}
    RESULTATS[nom] = r
    print(f"\nBilan de la {nom}")
    for cle, val in r.items():
        print(f"   {cle:<30} {val:6.1f}")

def manche_1(mon_offre):
    joueurs = distribuer()
    for j in joueurs:
        offre = min(mon_offre, j["v"]) if j["nom"] == "vous" else round(j["v"] * random.uniform(0.25, 0.75), 1)
        j["offre"] = offre
    ordre = sorted(joueurs, key=lambda j: (j["offre"], random.random()), reverse=True)
    for rang, j in enumerate(ordre):
        if rang < PLACES * BLOCS:
            j["bloc"], j["pourboire"] = rang // PLACES + 1, j["offre"]
    for b in range(1, BLOCS + 1):
        dans_b = [j for j in ordre if j["bloc"] == b]
        print(f"Bloc {b} : " + ", ".join(f"{'VOUS' if j['nom'] == 'vous' else j['type']} ({j['offre']:g})" for j in dans_b))
    print("Exclu.es : " + ", ".join("VOUS" if j["nom"] == "vous" else j["type"] for j in ordre if not j["bloc"]))
    moi = next(j for j in joueurs if j["nom"] == "vous")
    print(f"\nVotre offre : {moi['offre']:g}. " + (f"Vous passez dans le bloc {moi['bloc']} et gagnez {gain(moi):.1f} points."
                                                   if moi["bloc"] else "Vous êtes exclu.e et gagnez 0."))
    bilan(joueurs, "manche 1")

def nouvelle_manche(choc=False):
    global etat
    etat = {"bloc": 0, "base": 2, "joueurs": distribuer(choc), "choc": choc}
    for j in etat["joueurs"]:
        j["part"] = random.uniform(0.3, 0.7)     # part du gain que chacun.e accepte de mettre en pourboire
    print(f"Nouvelle manche{' avec choc de demande (10 cartes NFT)' if choc else ''}. Frais de base du bloc 1 : {etat['base']} points.")

def bloc_suivant(je_paie_le_frais_de_base, mon_pourboire_max=0):
    if etat["bloc"] >= BLOCS:
        print("La manche est finie. Pour la rejouer, relancez la cellule qui appelle nouvelle_manche.")
        return
    etat["bloc"] += 1
    b, base = etat["bloc"], etat["base"]
    candidats = []
    for j in etat["joueurs"]:
        if j["bloc"]:
            continue
        reste = j["v"] - j["w"] * (b - 1) - base
        if j["nom"] == "vous":
            if je_paie_le_frais_de_base and reste >= 0:
                j["max"] = min(mon_pourboire_max, reste); candidats.append(j)
        elif reste >= 1:
            j["max"] = int(reste * j["part"]); candidats.append(j)
    candidats.sort(key=lambda j: (j["max"], random.random()), reverse=True)
    pourboire = candidats[PLACES]["max"] if len(candidats) > PLACES else 0
    inclus = candidats[:PLACES]
    for j in inclus:
        j.update(bloc=b, base=base, pourboire=pourboire)
    print(f"Bloc {b} : frais de base {base}, {len(candidats)} candidat.es pour {PLACES} places, pourboire {pourboire}.")
    print("   Inclus : " + (", ".join("VOUS" if j["nom"] == "vous" else j["type"] for j in inclus) or "personne"))
    moi = next(j for j in etat["joueurs"] if j["nom"] == "vous")
    if moi["bloc"] == b:
        print(f"   Votre transaction passe. Votre gain : {gain(moi):.1f} points.")
    elif not moi["bloc"]:
        print("   Votre transaction attend.")
    if len(inclus) == PLACES:
        etat["base"] = max(1, round(base * 1.25, 1))
    elif len(inclus) <= PLACES / 2:
        etat["base"] = max(1, round(base * 0.8, 1))
    if b < BLOCS:
        print(f"   Frais de base annoncé pour le bloc {b + 1} : {etat['base']}.")
    else:
        bilan(etat["joueurs"], "manche 3" if etat["choc"] else "manche 2")

print(f"Votre carte : {ma_carte}, v = {v_moi}, w = {w_moi}.")
print(f"Votre gain si vous passez dans le bloc 1 sans frais : {v_moi} points. Chaque bloc d'attente vous coûte {w_moi} point(s).")
```

**Ce que vous allez faire.** Simuler deux règles de frais sur 160 blocs, avec un choc de demande entre les blocs 60 et 100. Dans le premier cas, chaque utilisateur.rice devine le frais à offrir dans une enchère au premier prix. Dans le second, la règle EIP-1559 affiche un frais de base qui s'ajuste. Le modèle est une simplification construite pour l'exercice.

<div style="border-left:4px solid #1F3A5F;background-color:#F1F3F6;color:#1E293B;padding:8px 16px 4px 16px;margin:16px 0 18px 0">

<b style="color:#1F3A5F">Rappel du chapitre</b>

Dans une **enchère au premier prix**, chacun.e paie son offre et doit deviner l'offre minimale qui suffit pour entrer dans le prochain bloc. Avec la règle **EIP-1559**, le frais de base augmente d'au plus 12,5 % quand le bloc précédent était rempli au-delà de la moitié de sa capacité maximale et baisse sinon. Il est détruit. Un petit pourboire revient au producteur du bloc.

<div style="border-left:4px solid #475569;background-color:#F3F4F6;color:#1E293B;padding:8px 16px 4px 16px;margin:16px 0 18px 0">

<b style="color:#475569">Les règles du modèle simplifié</b>

1. À chaque bloc, 9 nouvelles transactions arrivent en moyenne, 27 pendant le choc. Chaque transaction a une valeur tirée au hasard, de 10 points en moyenne.
2. **Enchère au premier prix.** La capacité est de 10 transactions par bloc. Chacun.e offre le prix d'entrée du bloc précédent multiplié par un nombre tiré entre 0,7 et 1,5, sans dépasser sa valeur. Les 10 offres les plus élevées passent.
3. **EIP-1559.** La capacité maximale est de 20 transactions et la cible de 10. Les transactions dont la valeur couvre le frais de base et un pourboire de 0,1 point sont candidates. Si elles sont plus de 20, celles de plus haute valeur passent.
4. Une transaction qui attend plus de 20 blocs est abandonnée.

</div>

<span style="color:#4A7C59;font-weight:bold">&#9660; La cellule ci-dessous</span> simule les deux règles avec les mêmes arrivées de transactions et trace, bloc par bloc, le frais moyen payé et le nombre de transactions incluses.

```python
import math, random
import matplotlib.pyplot as plt
COULEURS = ["#1F3A5F", "#4A7C59", "#B45309", "#9F1239"]   # couleurs du cours

def arrivees(nb_blocs=160, debut_choc=60, fin_choc=100, graine=1):
    """Les mêmes transactions pour les deux règles : liste, bloc par bloc, de leurs valeurs."""
    random.seed(graine)
    return [[random.expovariate(1 / 10) for _ in range(tirage_poisson(27 if debut_choc <= b < fin_choc else 9))]
            for b in range(nb_blocs)]

def premier_prix(flux, capacite=10):
    file, prix_entree, frais, nombre, surpaiement = [], 1.0, [], [], []
    for b, nouvelles in enumerate(flux):
        file += [{"v": v, "arrivee": b} for v in nouvelles]
        for tx in file:
            tx["offre"] = min(tx["v"], max(0.1, prix_entree * random.uniform(0.7, 1.5)))
        file.sort(key=lambda tx: tx["offre"], reverse=True)
        inclus, file = file[:capacite], file[capacite:]
        file = [tx for tx in file if b - tx["arrivee"] < 20]
        if inclus:
            prix_entree = min(tx["offre"] for tx in inclus)
            surpaiement.append(sum(tx["offre"] - prix_entree for tx in inclus) / len(inclus))
        frais.append(sum(tx["offre"] for tx in inclus) / max(len(inclus), 1))
        nombre.append(len(inclus))
    return frais, nombre, sum(surpaiement) / len(surpaiement)

def eip1559(flux, cible=10, maximum=20, ajustement=0.125, pourboire=0.1):
    file, base, frais, nombre = [], 1.0, [], []
    for b, nouvelles in enumerate(flux):
        file += [{"v": v, "arrivee": b} for v in nouvelles]
        candidats = sorted((tx for tx in file if tx["v"] >= base + pourboire), key=lambda tx: tx["v"], reverse=True)
        inclus = candidats[:maximum]
        file = [tx for tx in file if tx not in inclus and b - tx["arrivee"] < 20]
        frais.append(base + pourboire if inclus else 0)
        nombre.append(len(inclus))
        base = base * (1 + ajustement * (len(inclus) - cible) / cible)   # +12,5 % au plus si le bloc est plein
    return frais, nombre

flux = arrivees()
frais_pp, nombre_pp, surpaiement_pp = premier_prix(flux)
frais_eip, nombre_eip = eip1559(flux)
print(f"Surpaiement moyen dans l'enchère au premier prix : {surpaiement_pp:.2f} point par transaction")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 3.8))
ax1.plot(frais_pp, color=COULEURS[2], label="enchère au premier prix")
ax1.plot(frais_eip, color=COULEURS[0], label="EIP-1559")
ax1.set_ylabel("frais moyen payé (points)")
ax2.plot(nombre_pp, color=COULEURS[2], label="enchère au premier prix")
ax2.plot(nombre_eip, color=COULEURS[0], label="EIP-1559")
ax2.set_ylabel("transactions incluses par bloc")
for ax, titre in ((ax1, "Frais payés"), (ax2, "Remplissage des blocs")):
    ax.axvspan(60, 100, color=COULEURS[1], alpha=0.12)
    ax.set_xlabel("bloc")
    ax.set_title(titre, color="#1F3A5F", loc="left")
    ax.spines[["top", "right"]].set_visible(False)
    ax.grid(True, alpha=0.25)
ax1.legend(frameon=False, fontsize=8)
plt.show()
```
